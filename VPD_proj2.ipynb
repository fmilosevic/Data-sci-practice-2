# Korelace


```python
num_col= df.select_dtypes(include=['number'])


corr = num_col.corr()
corr
```

```python
upper_triangle = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))


sorted_corr = upper_triangle.stack().sort_values(ascending=False)

sorted_corr

```

```python
plt.figure(figsize=(14, 6))
plt.plot(sorted_corr.values, marker='o')
plt.grid(True)
```

```python
High_corr = sorted_corr[sorted_corr.notna() & (1.00 >= sorted_corr) & (sorted_corr >= 0.95)]
High_corr
```

```python
High_negativ_corr = sorted_corr[sorted_corr.notna() & (sorted_corr <= -0.70)]
High_negativ_corr
```

# Turbo

```python
turbo = df[df['turbocharger'].notna()]


non_turbo = df[df['turbocharger'].isna()]


mean_turbo = turbo['annual_consumption_in_barrels_ft1'].mean()
mean_nonturbo = non_turbo['annual_consumption_in_barrels_ft1'].mean()


print("turbo consumption: ", mean_turbo)
print("non-turbo consumption: ", mean_nonturbo)
turbo_diff = mean_nonturbo/mean_turbo
print("turbo diff %: ", turbo_diff)
```

```python
turbo_count = df['turbocharger'].notna().sum()


non_turbo_count = df['turbocharger'].isna().sum()

print("turbochargers: ", turbo_count)
print("non-turbochargers: ", non_turbo_count)
turbo_count_diff = turbo_count/(non_turbo_count+turbo_count)
print("% of turbos : ", turbo_count_diff*100)
```

```python

super = df['supercharger'].notna().sum()


non_super = df['supercharger'].isna().sum()

print("superchargers: ", super)
print("non-superchargers: ", non_super)
```

# Prevodovka

```python
fuel_eff_trans = df.groupby('transmission')['combined_mpg_ft1'].mean()
fuel_eff_trans
```

```python
def transmiss_combine(tr):
    if pd.isna(tr):
        return np.nan
    elif 'Auto' in tr or 'Automatic' in tr:
        return 'Automatic'
    elif 'Manual' in tr or 'M' in tr:
        return 'Manual'
    else:
        return 'Other'


df['transmiss_combined'] = df['transmission'].apply(transmiss_combine)

test_count = df['transmiss_combined'].value_counts()



transmiss_categ = df.groupby('transmiss_combined')['combined_mpg_ft1'].mean()


transmiss_categ
```
